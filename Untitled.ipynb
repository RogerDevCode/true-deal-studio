```python
import pandas as pd

bd = pd.read_excel("banco.xlsx")

# ej1: cuantos clientes hay en cada sucursal
print("Ejercicio 1: cantidad de clientes por sucursal")
csuc = bd.groupby("Sucursal")["Edad"].count()
print(csuc)
smay = csuc.idxmax()
smen = csuc.idxmin()
print("sucursal con mas clientes:", smay, "->", csuc.max())
print("sucursal con menos clientes:", smen, "->", csuc.min())
print()

# ej2: cuantos clientes por sexo
print("Ejercicio 2: cantidad de clientes por sexo")
csex = bd.groupby("Sexo")["Edad"].count()
print(csex)
sxmay = csex.idxmax()
dsex = csex.max() - csex.min()
print("sexo con mas clientes:", sxmay)
print("diferencia entre sexos:", dsex)
print()

# ej3: ingreso promedio por sucursal
print("Ejercicio 3: ingreso anual promedio por sucursal")
ipsuc = bd.groupby("Sucursal")["Ingreso_Anual"].mean()
print(ipsuc)
simay = ipsuc.idxmax()
simen = ipsuc.idxmin()
print("sucursal con mayor ingreso promedio:", simay)
print("sucursal con menor ingreso promedio:", simen)
print()

# ej4: ingreso promedio por sexo
print("Ejercicio 4: ingreso anual promedio por sexo")
ipsex = bd.groupby("Sexo")["Ingreso_Anual"].mean()
print(ipsex)
sximay = ipsex.idxmax()
disex = ipsex.max() - ipsex.min()
print("sexo con mayor ingreso promedio:", sximay)
print("diferencia de ingreso entre sexos:", disex)
print()

# ej5: edad promedio por sucursal
print("Ejercicio 5: edad promedio por sucursal")
epsuc = bd.groupby("Sucursal")["Edad"].mean()
print(epsuc)
seamay = epsuc.idxmax()
desuc = epsuc.max() - epsuc.min()
print("sucursal con clientes de mayor edad:", seamay)
print("diferencia de edad entre sucursales:", desuc)
print()

# ej6: edad promedio por sexo
print("Ejercicio 6: edad promedio por sexo")
epsex = bd.groupby("Sexo")["Edad"].mean()
print(epsex)
sxemay = epsex.idxmax()
print("sexo con mayor edad promedio:", sxemay)
print()

# ej7: ingreso maximo por sucursal
print("Ejercicio 7: ingreso anual maximo por sucursal")
imsuc = bd.groupby("Sucursal")["Ingreso_Anual"].max()
print(imsuc)
smax = imsuc.idxmax()
print("sucursal con el ingreso mas alto:", smax, "->", imsuc.max())
print()

# ej8: ingreso minimo por sucursal
print("Ejercicio 8: ingreso anual minimo por sucursal")
iminsuc = bd.groupby("Sucursal")["Ingreso_Anual"].min()
print(iminsuc)
smin = iminsuc.idxmin()
print("sucursal con el ingreso mas bajo:", smin, "->", iminsuc.min())
dmm = imsuc - iminsuc
print("diferencia entre ingreso max y min por sucursal:")
print(dmm)
print()

# ej9: ingreso promedio cruzando sexo y sucursal
print("Ejercicio 9: ingreso anual promedio segun sexo y sucursal")
isxsuc = bd.groupby(["Sexo", "Sucursal"])["Ingreso_Anual"].mean()
print(isxsuc)
cmay = isxsuc.idxmax()
cmen = isxsuc.idxmin()
print("combinacion con mayor ingreso promedio:", cmay)
print("combinacion con menor ingreso promedio:", cmen)
print()

# ej10: clientes con tarjeta por sucursal
print("Ejercicio 10: cantidad de clientes con tarjeta por sucursal")
ctarj = bd[bd["Tarjetas"] == "SI"]
tsuc = ctarj.groupby("Sucursal")["Edad"].count()
print(tsuc)
stmay = tsuc.idxmax()
stmen = tsuc.idxmin()
print("sucursal con mas clientes con tarjeta:", stmay)
print("sucursal con menos clientes con tarjeta:", stmen)
print()

# desafio: tabla resumen con todos los indicadores por sucursal
print("Desafio: tabla resumen por sucursal")
tab = bd.groupby("Sucursal").agg(
    cant=("Edad", "count"),
    edadprom=("Edad", "mean"),
    ingprom=("Ingreso_Anual", "mean"),
    ingmax=("Ingreso_Anual", "max"),
    ingmin=("Ingreso_Anual", "min")
)
print(tab)
scamp = tab["ingprom"].idxmax()
smcli = tab["cant"].idxmax()
print("sucursal con mayor ingreso promedio:", scamp)
print("sucursal con mas clientes:", smcli)
print("para una campana de clientes de altos ingresos me enfocaria en la sucursal", scamp, "porque ahi los clientes ganan mas en promedio")
```
